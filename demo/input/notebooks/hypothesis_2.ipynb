# Гипотеза 2

Генеральная дисперсия подготовленного признака `value` меньше `10`.

```python
import csv
import json
import os
import statistics
from pathlib import Path

with Path("data/prepared.csv").open(newline="", encoding="utf-8") as stream:
    values = [float(row["value"]) for row in csv.DictReader(stream)]

metric = statistics.pvariance(values)
threshold = 10.0
accepted = metric < threshold

output = (
    Path("runs")
    / os.environ["MLDEV_RUN_ID"]
    / "hypothesis_2"
    / "result.json"
)
output.parent.mkdir(parents=True, exist_ok=True)
output.write_text(
    json.dumps(
        {
            "hypothesis": "h2",
            "metric": "population_variance",
            "value": metric,
            "threshold": threshold,
            "accepted": accepted,
        },
        indent=2,
    )
    + "\n",
    encoding="utf-8",
)

print(f"STAGE_RESULT hypothesis_2 accepted={accepted} output={output}")
```
