# Гипотеза 1

Среднее значение подготовленного признака `value` больше `5`.

```python
import csv
import json
import os
from pathlib import Path

with Path("data/prepared.csv").open(newline="", encoding="utf-8") as stream:
    values = [float(row["value"]) for row in csv.DictReader(stream)]

metric = sum(values) / len(values)
threshold = 5.0
accepted = metric > threshold

output = (
    Path("runs")
    / os.environ["MLDEV_RUN_ID"]
    / "hypothesis_1"
    / "result.json"
)
output.parent.mkdir(parents=True, exist_ok=True)
output.write_text(
    json.dumps(
        {
            "hypothesis": "h1",
            "metric": "mean",
            "value": metric,
            "threshold": threshold,
            "accepted": accepted,
        },
        indent=2,
    )
    + "\n",
    encoding="utf-8",
)

print(f"STAGE_RESULT hypothesis_1 accepted={accepted} output={output}")
```
